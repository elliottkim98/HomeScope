```python
import pandas as pd
import numpy as np

# --- 1. Load Datasets ---
# Assuming your raw files are named accordingly
acs_raw = pd.read_csv('/content/ACSST5Y2024.S1901-Data.csv') # Replace with actual filename
zhvi_raw = pd.read_csv('/content/City_zhvi_uc_sfr_tier_0.33_0.67_sm_sa_month.csv') # Replace with actual filename

def clean_and_process(acs_raw, zhvi_raw):
    # --- 2. Clean ACS Data ---
    # Rename columns based on metadata
    acs_cols = {
        'GEO_ID': 'geo_id',
        'NAME': 'full_name',
        'S1901_C01_001E': 'num_households',
        'S1901_C01_012E': 'household_income'
    }
    acs = acs_raw[list(acs_cols.keys())].rename(columns=acs_cols)

    # Extract City/County and State from full_name (e.g., "Philadelphia city, Pennsylvania")
    # This regex splits into [Name] and [State]
    split_names = acs['full_name'].str.split(', ', expand=True)
    acs['region_name_raw'] = split_names[0]
    acs['state_full'] = split_names[1]

    # State mapping (Full Name to Abbreviation for matching with Zillow)
    state_map = {
        'Alabama': 'AL', 'Alaska': 'AK', 'Arizona': 'AZ', 'Arkansas': 'AR', 'California': 'CA',
        'Colorado': 'CO', 'Connecticut': 'CT', 'Delaware': 'DE', 'Florida': 'FL', 'Georgia': 'GA',
        'Hawaii': 'HI', 'Idaho': 'ID', 'Illinois': 'IL', 'Indiana': 'IN', 'Iowa': 'IA',
        'Kansas': 'KS', 'Kentucky': 'KY', 'Louisiana': 'LA', 'Maine': 'ME', 'Maryland': 'MD',
        'Massachusetts': 'MA', 'Michigan': 'MI', 'Minnesota': 'MN', 'Mississippi': 'MS',
        'Missouri': 'MO', 'Montana': 'MT', 'Nebraska': 'NE', 'Nevada': 'NV', 'New Hampshire': 'NH',
        'New Jersey': 'NJ', 'New Mexico': 'NM', 'New York': 'NY', 'North Carolina': 'NC',
        'North Dakota': 'ND', 'Ohio': 'OH', 'Oklahoma': 'OK', 'Oregon': 'OR', 'Pennsylvania': 'PA',
        'Rhode Island': 'RI', 'South Carolina': 'SC', 'South Dakota': 'SD', 'Tennessee': 'TN',
        'Texas': 'TX', 'Utah': 'UT', 'Vermont': 'VT', 'Virginia': 'VA', 'Washington': 'WA',
        'West Virginia': 'WV', 'Wisconsin': 'WI', 'Wyoming': 'WY', 'District of Columbia': 'DC'
    }
    acs['state_abbr'] = acs['state_full'].map(state_map)

    # Standardize names: lowercase and remove common suffixes to help matching
    def normalize_name(name):
        if not isinstance(name, str): return name
        return name.lower().replace(' city', '').replace(' county', '').replace(' town', '').strip()

    acs['match_key'] = acs['region_name_raw'].apply(normalize_name) + "_" + acs['state_abbr'].fillna('')

    # --- 3. Clean and Pivot ZHVI Data ---
    # ZHVI is wide (dates as columns). Melt to long format.
    id_vars = ['RegionID', 'RegionName', 'StateName', 'RegionType'] # Adjust if columns differ
    # Identification of date columns (usually YYYY-MM-DD)
    date_cols = [c for c in zhvi_raw.columns if '-' in c and len(c) >= 7]

    zhvi_long = zhvi_raw.melt(id_vars=id_vars, value_vars=date_cols, var_name='date', value_name='value')
    zhvi_long['match_key'] = zhvi_long['RegionName'].apply(normalize_name) + "_" + zhvi_long['StateName']

    # --- 4. Create the Unified REGION Table ---
    # We use Zillow's RegionID as the primary key
    region_table = zhvi_raw[['RegionID', 'RegionName', 'StateName', 'RegionType']].copy()
    region_table = region_table.rename(columns={
        'RegionID': 'region_id',
        'RegionName': 'region_name',
        'StateName': 'state_name',
        'RegionType': 'region_type'
    })

    # --- 5. Final Table Preparation ---
    # Map ACS to region_id via match_key
    id_map = zhvi_long[['match_key', 'RegionID']].drop_duplicates().groupby('match_key')['RegionID'].first()
    acs['region_id'] = acs['match_key'].map(id_map)

    # Drop rows that couldn't be matched (Entity Resolution filter)
    acs_clean = acs.dropna(subset=['region_id', 'household_income'])
    acs_final = acs_clean[['region_id', 'num_households', 'household_income']].copy()
    acs_final['year'] = 2023 # Based on your metadata

    zhvi_final = zhvi_long[['RegionID', 'date', 'value']].rename(columns={'RegionID': 'region_id'})
    zhvi_final = zhvi_final.dropna(subset=['value'])

    return region_table, acs_final, zhvi_final

region_df, acs_df, zhvi_df = clean_and_process(acs_raw, zhvi_raw)
```

```python
import pandas as pd
import numpy as np

# --- 1. Load Datasets ---
# Assuming your raw files are named accordingly
acs_raw = pd.read_csv('/content/acs_clean_full_2010_2024.csv', skiprows=[1]) # Replace with actual filename
zhvi_raw = pd.read_csv('/content/City_zhvi_uc_sfr_tier_0.33_0.67_sm_sa_month.csv') # Replace with actual filename

def clean_and_process(acs_raw, zhvi_raw):
    # --- 1. Clean ACS Data ---
    # Based on your provided schema: geo_id, full_name, etc.
    # --- 1. Process ACS Data (Already Cleaned) ---
    # We just need to ensure data types are correct for the DB
    acs = acs_raw.copy()

    # Ensure numeric types
    acs['num_households'] = pd.to_numeric(acs['num_households'], errors='coerce')
    acs['household_income'] = pd.to_numeric(acs['household_income'], errors='coerce')

    # --- 2. Clean and Pivot ZHVI Data ---
    # Zillow data is usually 'wide' with dates as columns
    id_vars = ['RegionID', 'RegionName', 'State', 'RegionType']

    # Find date columns (e.g., 2024-01-31)
    date_cols = [c for c in zhvi_raw.columns if any(char.isdigit() for char in c) and '-' in c]

    zhvi_long = zhvi_raw.melt(id_vars=id_vars, value_vars=date_cols, var_name='date', value_name='value')

    # Standardize Zillow names to create a match_key compatible with your ACS match_key
    def normalize_name(name):
        if not isinstance(name, str): return ""
        name = name.lower()
        for suffix in [' city', ' town', ' cdp', ' village', ' borough']:
            name = name.replace(suffix, '')
        return name.strip()

    zhvi_long['match_key'] = zhvi_long['RegionName'].apply(normalize_name) + "_" + zhvi_long['State']

    # --- 3. Create the Unified REGION Table ---
    # This acts as your source of truth for geography
    region_table = zhvi_raw[id_vars].copy().rename(columns={
        'RegionID': 'region_id',
        'RegionName': 'region_name',
        'State': 'state_abbr',
        'RegionType': 'region_type'
    })

    # --- 4. Map ACS to region_id via the pre-existing match_key ---
    # We use Zillow's RegionID as the Primary Key for the DB
    id_map = zhvi_long.drop_duplicates('match_key').set_index('match_key')['RegionID']
    acs['region_id'] = acs['match_key'].map(id_map)

    # --- 5. Final Table Preparation ---
    # Filter for successful matches
    acs_final = acs.dropna(subset=['region_id']).copy()

    # Structure for DB: include the region_id and the metrics
    acs_final = acs_final[['region_id', 'year', 'num_households', 'household_income']]

    zhvi_final = zhvi_long[['RegionID', 'date', 'value']].rename(columns={'RegionID': 'region_id'})
    zhvi_final = zhvi_final.dropna(subset=['value'])

    return region_table, acs_final, zhvi_final

# Usage
region_df, acs_df, zhvi_df = clean_and_process(acs_raw, zhvi_raw)
```

```python
#write datasets to csv
region_df.to_csv('region_df.csv', index=False)
acs_df.to_csv('acs_df.csv', index=False)
zhvi_df.to_csv('zhvi_df.csv', index=False)
```
